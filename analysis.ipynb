```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv("helicopter_prison_escapes.csv")

# Show basic information
print("Total records:", len(df))
df.head()
```

```python
# Count of escapes by country
country_counts = df['Country'].value_counts()
print(country_counts)

# Plotting the number of escapes by country
country_counts.plot(kind='bar', figsize=(12,6))
plt.title("Number of Helicopter Prison Escapes by Country")
plt.xlabel("Country")
plt.ylabel("Number of Escapes")
plt.tight_layout()
plt.show()
```

```python
# Count of successful vs failed escapes
success_counts = df['Succeeded'].value_counts()
print(success_counts)

# Plotting success vs failure
success_counts.plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
plt.title("Success vs Failure in Helicopter Prison Escapes")
plt.ylabel("")  # Remove default label
plt.show()
```

```python
# Convert 'Date' to datetime format
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Plot escapes over time
escapes_over_time = df.groupby(df['Date'].dt.year)['Date'].count()
print(escapes_over_time)

escapes_over_time.plot(kind='line', marker='o', figsize=(10,5))
plt.title("Helicopter Prison Escapes Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Escapes")
plt.grid(True)
plt.show()
```
